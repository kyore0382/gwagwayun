```python
from sleap_nn.config.get_config import (
    get_data_config,
    get_model_config,
    get_trainer_config
)
from omegaconf import OmegaConf

# 데이터 설정
data_config = get_data_config(
    train_labels_path=["/content/drive/MyDrive/sleap/colab.pkg.slp"],
    scale=0.5,
    crop_size=100,
    user_instances_only=True
)

# 모델 설정
model_config = get_model_config(
    backbone_config="unet",
    head_configs="centroid"
)

# 학습 설정
trainer_config = get_trainer_config(
    batch_size=2,
    max_epochs=50,
    trainer_accelerator="gpu",
    save_ckpt=True,
    ckpt_dir="/content/drive/MyDrive/sleap/model_new",
    learning_rate=1e-4
)

config = OmegaConf.create({
    "data_config": OmegaConf.structured(data_config),
    "model_config": OmegaConf.structured(model_config),
    "trainer_config": OmegaConf.structured(trainer_config)
})

print(OmegaConf.to_yaml(config))
```

```python
from sleap_nn.train import run_training

run_training(config)
```

```python
from omegaconf import OmegaConf
from sleap_nn.train import run_training
import torch

print("CUDA:", torch.cuda.is_available())
if torch.cuda.is_available():
    print(torch.cuda.get_device_name(0))

config = OmegaConf.load(
    "/content/drive/MyDrive/sleap/config.yaml"
)

config.trainer_config.trainer_accelerator = "gpu"
config.trainer_config.trainer_devices = 1

config.trainer_config.train_data_loader.num_workers = 2
config.trainer_config.val_data_loader.num_workers = 2

print("학습 시작")
run_training(config)
```
